## **Fraudulent Claim Detection**

## Problem Statement
Global Insure, a leading insurance company, processes thousands of claims annually. However, a significant percentage of these claims turn out to be fraudulent, resulting in considerable financial losses. The company’s current process for identifying fraudulent claims involves manual inspections, which is time-consuming and inefficient. Fraudulent claims are often detected too late in the process, after the company has already paid out significant amounts. Global Insure wants to improve its fraud detection process using data-driven insights to classify claims as fraudulent or legitimate early in the approval process. This would minimise financial losses and optimise the overall claims handling process.

## Business Objective
Global Insure wants to build a model to classify insurance claims as either fraudulent or legitimate based on historical claim details and customer profiles. By using features like claim amounts, customer profiles and claim types, the company aims to predict which claims are likely to be fraudulent before they are approved.

### **2.1 Handle null values** <font color = red>[2 marks]</font>

#### **2.1.2** Handle rows containing null values <font color="red">[1 Mark]</font>

```python
# Handle the rows containing null values

# 'authorities_contacted' is the only feature with missing values (91 rows), and they are
# not gaps in the record. The raw CSV holds the string "None" for these claims, which
# pandas reads as missing by default, so a null records a claim where no authority was
# contacted. Treating it as its own category keeps all 999 rows; dropping would cost 9%
# of a small dataset that only has 247 fraud cases to learn from.
raw_authorities = pd.read_csv("insurance_claims.csv", usecols=["authorities_contacted"],
                              keep_default_na=False)["authorities_contacted"]
print("Values as stored in the CSV:")
print(raw_authorities.value_counts())

print("\nCategories before:")
print(df["authorities_contacted"].value_counts(dropna=False))

df["authorities_contacted"] = df["authorities_contacted"].fillna("None")

print("\nCategories after:")
print(df["authorities_contacted"].value_counts(dropna=False))
print("\nRemaining nulls:", df["authorities_contacted"].isnull().sum())
```

### **2.2 Identify and handle redundant values and columns** <font color = red>[5 marks]</font>

#### **2.2.1** Examine the columns to determine if any value or column needs to be treated <font color="red">[2 Mark]</font>

```python
# Write code to display all the columns with their unique values and counts and check for redundant values
for col in df.select_dtypes(include="object").columns:
    print(f"--- {col}  ({df[col].nunique()} unique) ---")
    print(df[col].value_counts())
    print()

# '?' is used as a placeholder in collision_type, property_damage and
# police_report_available. It is kept as its own level rather than imputed.
# Columns worth noting:
#   - auto_model (39 levels) and insured_hobbies (20 levels) are very granular and will
#     be grouped in section 6.4
#   - incident_location has a unique value per row and is handled in 2.2.4
#   - policy_csl holds strings like "250/500", a coverage-limit ratio rather than a
#     number, so it stays categorical
```

#### **2.2.2** Identify and drop any columns that are completely empty <font color="red">[1 Mark]</font>

```python
# Identify and drop any columns that are completely empty
empty_cols = df.columns[df.isnull().all()].tolist()
print("Completely empty columns:", empty_cols)

df = df.drop(columns=empty_cols)
print("Shape after drop:", df.shape)

# '_c39' is an artefact of a trailing comma in the CSV export - 1000 nulls, no values.
```

#### **2.2.3** Identify and drop rows where features have illogical or invalid values, such as negative values for features that should only have positive values <font color="red">[1 Mark]</font>

```python
# Identify and drop rows where features have illogical or invalid values, such as negative values for features that should only have positive values

# 'capital-loss' is deliberately excluded: it is negative by definition, so including it
# would delete several hundred valid rows.
must_be_positive = ["umbrella_limit", "total_claim_amount", "injury_claim",
                    "property_claim", "vehicle_claim", "policy_annual_premium"]

for col in must_be_positive:
    n_bad = (df[col] < 0).sum()
    if n_bad:
        print(f"{col}: {n_bad} negative value(s), min = {df[col].min():,}")

invalid = (df[must_be_positive] < 0).any(axis=1)
print(f"\nDropping {invalid.sum()} row(s)")

df = df.loc[~invalid].reset_index(drop=True)
print("Shape after drop:", df.shape)

# A single claim carries umbrella_limit = -1,000,000. A negative coverage ceiling is
# not possible, so the row is removed rather than repaired - one row out of a thousand
# is not worth guessing the intended sign for.
```

#### **2.2.4** Identify and remove columns where a large proportion of the values are unique or near-unique, as these columns are likely to be identifiers or have very limited predictive power <font color="red">[1 Mark]</font>

```python
# Identify and remove columns that are likely to be identifiers or have very limited predictive power
uniqueness = (df.nunique() / len(df)).sort_values(ascending=False)
print("Proportion of unique values, highest first:")
print(uniqueness.head(10).round(3))

# Before dropping the date columns, derive the features that need them. The gap between
# policy start and incident is plausibly informative: a claim shortly after a policy is
# bound is a recognised fraud signal in insurance.
df["policy_bind_date"] = pd.to_datetime(df["policy_bind_date"])
df["incident_date"] = pd.to_datetime(df["incident_date"])
df["days_policy_to_incident"] = (df["incident_date"] - df["policy_bind_date"]).dt.days
df["vehicle_age"] = df["incident_date"].dt.year - df["auto_year"]

identifiers = ["policy_number", "insured_zip", "incident_location"]
df = df.drop(columns=identifiers + ["policy_bind_date", "incident_date", "auto_year"])

print("\nDropped:", identifiers, "+ raw date columns (replaced by derived features)")
print("Shape:", df.shape)

# policy_number and incident_location hold one distinct value per row, insured_zip 994
# of 999. A model can only memorise such columns, never generalise from them.
```

### **4.4 Perform bivariate analysis** <font color="red">[10 Marks]</font>

#### **4.4.1** Target likelihood analysis for categorical variables. <font color="red">[5 Marks]</font>
Investigate the relationships between categorical features and the target variable by analysing the target event likelihood (for the `'Y'` event) for each level of every relevant categorical feature. Through this analysis, identify categorical features that do not contribute much in explaining the variation in the target variable.

```python
# Write a function to calculate and analyse the target variable likelihood for categorical features
def target_likelihood(column, min_count=15):
    """Fraud rate and support for each level of a categorical feature."""
    frame = pd.DataFrame({"level": X_train[column].astype(str), "fraud": y_train.values})
    out = (frame.groupby("level")["fraud"]
           .agg(n="size", fraud_rate="mean")
           .sort_values("fraud_rate", ascending=False))
    out["fraud_rate"] = (out["fraud_rate"] * 100).round(1)
    return out[out["n"] >= min_count]


base_rate = y_train.mean() * 100
print(f"Base fraud rate: {base_rate:.1f}%\n")

# Rank features by how much the fraud rate varies across their levels
spread = []
for col in categorical_cols:
    t = target_likelihood(col)
    if len(t) > 1:
        spread.append((col, t["fraud_rate"].max(), t["fraud_rate"].min(),
                       t["fraud_rate"].max() - t["fraud_rate"].min()))

spread_df = (pd.DataFrame(spread, columns=["feature", "max_rate", "min_rate", "spread"])
             .sort_values("spread", ascending=False))
print(spread_df.to_string(index=False))
```

#### **4.4.2** Explore the relationships between numerical features and the target variable to understand their impact on the target outcome using appropriate visualisation techniques to identify trends and potential interactions. <font color="red">[5 Marks]</font>

```python
# Visualise the relationship between numerical features and the target variable to understand their impact

# The two strongest categorical signals first
chart_files = {"incident_severity": "images/fraud-rate-severity.png",
               "insured_hobbies": "images/fraud-rate-hobbies.png"}

for col in ["incident_severity", "insured_hobbies"]:
    t = target_likelihood(col)
    name = col.replace("_", " ")
    plt.figure(figsize=(9, max(3.5, 0.32 * len(t))))
    ax = sns.barplot(x=t["fraud_rate"], y=t.index, color="#4C72B0")
    ax.bar_label(ax.containers[0], padding=3, fontsize=8,
                 bbox=dict(facecolor="white", edgecolor="none", pad=0.8),
                 labels=[f"{r:.1f}% (n={n})" for r, n in zip(t["fraud_rate"], t["n"])])
    plt.axvline(base_rate, color="#C44E52", ls="--", label=f"base rate {base_rate:.1f}%")
    plt.xlim(0, t["fraud_rate"].max() * 1.22)
    plt.xlabel("Fraud rate (%)")
    plt.ylabel(name.capitalize())
    plt.title(f"Fraud rate by {name} (training set)")
    plt.legend(loc="lower right")
    plt.tight_layout()
    plt.savefig(chart_files[col], dpi=150, bbox_inches="tight")
    plt.show()
    print(t.to_string(), "\n")

# Then the numerical features against the target
key_numeric = ["total_claim_amount", "vehicle_claim", "injury_claim",
               "property_claim", "days_policy_to_incident", "age"]

plt.figure(figsize=(15, 6))
for i, col in enumerate(key_numeric):
    plt.subplot(2, 3, i + 1)
    lab = y_train.map({0: "No", 1: "Yes"})
    sns.boxplot(x=lab, y=X_train[col], hue=lab,
                palette=["#4C72B0", "#C44E52"], legend=False, fliersize=2)
    plt.title(col, fontsize=10)
    plt.xlabel("Fraud reported")
    plt.ylabel("")

plt.suptitle("Numerical features vs fraud (training set)", y=1.002, fontsize=13)
plt.tight_layout()
plt.show()
```

#### **7.2.3** Evaluate VIF of features to assess multicollinearity <font color="red">[2 Marks]</font>

```python
# Import 'variance_inflation_factor'
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
# Make a VIF DataFrame for all the variables present
def vif_frame(frame):
    out = pd.DataFrame({"feature": frame.columns})
    out["VIF"] = [variance_inflation_factor(frame.values, i) for i in range(frame.shape[1])]
    return out.sort_values("VIF", ascending=False).reset_index(drop=True)


vif_frame(X_train_rfecv).round(2)
```

Proceed to the next step if p-values and VIFs are within acceptable ranges. If you observe high p-values or VIFs, drop the features and retrain the model. <font color="red">[THIS IS OPTIONAL]</font>

#### **7.2.4** Make predictions on training data <font color = "red">[1 Mark]</font>

```python
# Predict the probabilities on the training data

# RFECV optimises predictive score, not statistical significance, so its picks can
# carry p-values above 0.05. Features are removed one at a time - always the
# least significant - refitting after each removal, and then any remaining feature with
# VIF above 5 is dropped the same way. Removing several at once is unsafe because each
# p-value is conditional on the rest of the model.
features = list(col)
removed = []

while len(features) > 5:
    fit = sm.GLM(y_train, sm.add_constant(X_train[features]),
                 family=sm.families.Binomial()).fit()
    pvals = fit.pvalues.drop("const")
    if pvals.max() > 0.05:
        worst = pvals.idxmax()
        removed.append((worst, "p-value", round(pvals.max(), 3)))
        features.remove(worst)
        continue

    v = vif_frame(X_train[features])
    if v["VIF"].iloc[0] > 5:
        worst = v["feature"].iloc[0]
        removed.append((worst, "VIF", round(v["VIF"].iloc[0], 2)))
        features.remove(worst)
        continue
    break

print(f"Removed {len(removed)} features:")
for f, why, val in removed:
    print(f"  {f:45s} {why} = {val}")

print(f"\nFinal model: {len(features)} features")

final_sm = sm.add_constant(X_train[features])
res = sm.GLM(y_train, final_sm, family=sm.families.Binomial()).fit()
print(res.summary())

print("\nHighest remaining p-value: %.4f" % res.pvalues.drop("const").max())
print("Highest remaining VIF:     %.2f" % vif_frame(X_train[features])["VIF"].max())

# Probabilities on the training data
y_train_pred = res.predict(final_sm)
y_train_pred[:10]
```

## **Evaluation and Conclusion**

My final model is the 6-feature logistic regression with a cutoff of 0.60. On the validation set, which keeps the natural fraud rate, it caught 62 of the 74 fraudulent claims, with sensitivity 0.838, specificity 0.850, precision 0.646, F1 0.729 and AUC 0.855.

The tuned random forest overfitted. It scored 0.982 accuracy on the training data but did worse than the logistic regression on every validation metric (F1 0.659, AUC 0.825). Its cross-validated F1 of 0.926 was also too optimistic, because I oversampled before cross-validation and copies of the same fraud claims ended up in both training and test folds. The validation set was never resampled, so its numbers are the ones I trust.

Most of the signal comes from incident severity and the high-risk hobby flag. Incident severity and insured hobbies had the two biggest spreads in fraud rate in the EDA, and in the final model the hobby flag and the three severity levels have the four largest coefficients. Severity makes sense, since Major Damage claims were fraudulent 60.9% of the time. The hobby flag does not: there is no real reason chess and cross-fit players would have fraud rates of 90.6% and 76.0%, so I think it is an artefact of the synthetic data and would not carry over to real claims.

The 0.60 cutoff treats a missed fraud and a false alarm as equally costly, which is not true for an insurer. The next step would be to set the cutoff from the real cost of each type of error, and to move the oversampling inside each cross-validation fold.
